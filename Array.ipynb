```python
import numpy as np

# Mathematical operations
marks = np.array([50, 60, 70, 80, 90])

print("Original Marks:", marks)
print("Adding +5 Marks:", marks + 5)

mul = marks * 100
print("Multiplying by 100:", mul)
```

```python
# mathamatical operation with two array
arr21=np.array([10,20,30])
arr22=np.array([1,2,3])
Addition =arr21+arr22
sub=arr21-arr22
mul=arr21*arr22
div=arr21/arr22
print("Addition",Addition)
print("Subtraction",sub)
print("Multiplication",mul)
print("Division",div)
```

```python
# Aggeregate Funaction
#sum
#min
#max
#avg
#count
arr31=np.array([45,67,78,89,56])
print(arr31)
print("Total:",arr31.sum)
print("\n Minimum Value:",arr31.min())
print("\n Maximum Value:",arr31.max())
print("\n Average Value:",arr31.mean())
```

```python
#Filtering function
arr41=np.array([35,67,42,89,25,76])
result=arr41[arr41>=50]
print(result)
```

```python
import numpy as np

# np.where
# to find position satisfying a condition

marks = np.array([35, 67, 42, 89, 25, 76])

position = np.where(marks >= 50)

print("Position:", position)
```

```python
# sorting
sub_marks=np.array([35, 67, 42, 89, 25, 76])
sorted_marks=np.sort(sub_marks)
print(sorted_marks)
```

```python
#cymulative submmation
import numpy as np

patients = np.array([10, 15, 20, 12])
cumulative_summation = patients.cumsum()

# Added the missing comma after the string
print("Daily Patients :", cumulative_summation)
print("\nPatient for 4 days:", cumulative_summation)
```
