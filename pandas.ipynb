```python
import pandas as pd
import numpy as np
```

```python
# Create your own dataset
data = {
    'student_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113],

    'name': ['Rahul kumar', 'amit sharma', 'PRIYA SINGH', 'Neha veram','Rvai verma', 'ravi kumar', 'Rvai Kumar', 'anjali gupta','vikas singh', 'POOJA SHARMA', 'Karan Metha', 'Arjun Singh','Riya Veram'],

    'Gender': ['Male', 'male', 'female', 'female', 'male', 'female', 'male','female', 'male', 'female', 'male', 'female', 'male'],

    'Department': ['cse', 'cse', 'ece', 'cse', 'ece', 'cse', 'ece','cse', 'me', 'cse', 'cse', 'ece', 'cse'],

    'age': [ 20, 21, np.nan, 75, 88, 65, 92, np.nan, 70, 95, 80, np.nan, 22],

    'Attendance': [85, 78, np.nan, 90 , 75, 88, 65, 92, np.nan, 70, 95, 80, np.nan],

    'Marks': [85, 72, np.nan, 75, 88, 65, 92, np.nan, 76, 64, 95, 81, 78 ],

    'Email': ['rahul@gmail.com', 'amit@gamil.com', 'priya@gmail','neha@gmail.com', 'None', 'anjali@gmail', 'vikas@gmail.com','pooja@gmail.com', 'vikas@gmail.com', 'pooja@gmail.com','karan@gmail.com', 'arjun@gmail.com', 'riya@gmail.com']
}
df = pd.DataFrame(data)
display(df)
```

```python
print('Shape:',df.shape)
print('columns:',df.columns.tolist)
print('Data Types ', df.dtypes)
display(df.head())
display(df.tail())
```

```python
 # Detect Missing values
print(df.isnull().sum())
```

```python
# Handle the missing values

df['age'] = df['age'].fillna(df['age'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Email'] = df['Email'].fillna('Not Provided')

# Check remaining missing values
print(df.isnull().sum())
```

```python
# data Trnsformation
df['Reslut ']=df['Marks'].apply(lambda x: 'Pass' if x >= 75 else 'Fail')
def assign_grade(marks):
    if marks >= 80:
        return 'A'
    elif marks >= 60:
        return 'B'
    elif marks >= 40:
        return 'C'
    else:
        return 'F'
df['Grade'] = df['Marks'].apply(assign_grade)
display(df)
```
