# 1. Array Creation & Properties

# 11. Integrated Lab Questions: Q39 Student Marks Analysis

### Q39

```python
np.random.seed(42)
marks_matrix = np.random.randint(30, 100, size=(10, 3))
totals = np.sum(marks_matrix, axis=1)
averages = np.mean(marks_matrix, axis=1)
passed_students = np.where(averages >= 50)[0]

print("Marks Matrix (10 Students x 3 Subjects):\n", marks_matrix)
print("Total Marks per Student:", totals)
print("Average Marks per Student:", averages)
print("Passed Student Indices (Avg >= 50):", passed_students)
```

# 11. Integrated Lab Questions: Q42 Matrix Operations Challenge

### Q42

```python
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Element-wise Addition:\n", A + B)
print("Element-wise Multiplication:\n", A * B)
matmul_res = np.matmul(A, B)
print("Matrix Multiplication (A @ B):\n", matmul_res)
print("Transpose of A:\n", A.T)
print("Max in MatMul Result:", np.max(matmul_res))
print("Min in MatMul Result:", np.min(matmul_res))
```

```python
arr_24 = np.arange(1, 25).reshape(4, 6)

# Extract sub-matrix using rows 0, 2 and columns 1, 3, 5
sub_mat = arr_24[[0, 2]][:, [1, 3, 5]]

print("Original 4x6 Array:\n", arr_24)
print("\nExtracted Sub-matrix:\n", sub_mat)
print("Transposed Shape:", arr_24.T.shape)
print("Row Means:", np.mean(arr_24, axis=1))
print("Column Means:", np.mean(arr_24, axis=0))
print("Unique Values:", np.unique(arr_24))
```
