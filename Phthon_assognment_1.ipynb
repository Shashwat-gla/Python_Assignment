```python
def my_abs(val):
    if isinstance(val, str):
        return "Invalid input: string cannot be compared"

    if val < 0:
        return 0 - val
    return val

print(my_abs("Hi"))
```

```python
# Q7.3 - Function prints absolute value instead of returning it

def print_abs(val):

    if val < 0:
        print(0 - val)
    else:
        print(val)


x = print_abs(-2.7)

print(x)
```

```python
# Q7.4 - Reassigning a local variable inside a function

def inc_val(val):
    val = val + 1


x = 7

inc_val(x)

print(x)
```

```python
# Q7.5 - Swap two local variables inside a function

def swap_message(val1, val2):

    val1, val2 = val2, val1

    print("Inside function:")
    print(val1)
    print(val2)


x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print(x)
print(y)
```

```python
# Q8.3 - Trying to modify a global variable without the global keyword

counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```

```python
# Q8.4 - Correctly modifying a global variable using the global keyword

counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
